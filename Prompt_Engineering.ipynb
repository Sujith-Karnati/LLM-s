Hello! Welcome to today's lesson. You already know about variables and data types, which is fantastic! Today, we are going to learn about one of the most useful tools in programming: **Functions**.

---

### 1. What is a Function?

Imagine you have a recipe to make a cup of coffee. Instead of writing down every single step (boil water, put coffee in mug, pour water, stir) every time you want coffee, you just say: **"Make Coffee."**

In Python, a **function** is like that recipe. It is a saved block of code that does a specific task.

Why do we use functions?
* **Reusability:** You write the code once, and you can use it as many times as you want.
* **Organization:** It keeps your code clean and easy to read.

---

### 2. Parameters and Return Values

To make functions even more useful, we use **Parameters** and **Return Values**. Think of a function like a juice blender:

* **Parameters (Inputs):** This is the information you put *into* the function. (Like putting apples into a blender).
* **Return Value (Output):** This is the result the function gives *back* to you when it finishes. (Like the apple juice coming out of the blender).

---

### 3. Code Examples

Let's look at two simple examples.

#### Example 1: A Greeting Function (Using Parameters)

```python
def say_hello(name):
    print("Hello, " + name + "! Welcome to Python.")


# Calling the function
say_hello("Alex")
say_hello("Sam")
```

**Explanation:**
1. **`def`**: This keyword tells Python, *"I am creating (defining) a function."*
2. **`say_hello`**: This is the name we chose for our function.
3. **`(name)`**: `name` is the **parameter** (the input). It holds whatever value we pass into the function.
4. **`say_hello("Alex")`**: This is how we "call" or run the function. We pass `"Alex"` as the input, so Python prints: `Hello, Alex! Welcome to Python.`

---

#### Example 2: Adding Numbers (Using Parameters and a Return Value)

```python
def add_two_numbers(num1, num2):
    total = num1 + num2
    return total


# Calling the function and saving the result
result = add_two_numbers(5, 3)
print(result)
```

**Explanation:**
1. **`num1, num2`**: This function takes *two* **parameters** (inputs).
2. **`total = num1 + num2`**: The function adds the two numbers together.
3. **`return total`**: The **`return`** keyword tells Python to send the final answer back to us.
4. **`result = add_two_numbers(5, 3)`**: We pass `5` and `3` into the function. The function calculates `8` and returns it. We store that `8` inside the variable named `result` and print it.

---

### Quick Summary

* **Function:** A reusable box of code (`def function_name():`).
* **Parameter:** The input you give to the function.
* **Return Value:** The output the function sends back to you using the `return` keyword.

# Chain of thought prompting

````python
prompt ="""
Solve the following problem step by step.

A shop sells a laptop for $800.
The shop offers a 15% discount.
What is the final price after applying the discount?

Explain the calculation clearly and provide the final answer.

"""

response = llm.models.generate_content(model="gemini-3.5-flash-lite",
contents=prompt)

print(response.text)
````

To find the final price of the laptop after applying the discount, we can follow these steps:

**Step 1: Understand the given information**
* Original price of the laptop = $800
* Discount percentage = 15%

**Step 2: Calculate the discount amount**
To find out how much money is being taken off, calculate 15% of the original price ($800).
You can do this by converting the percentage to a decimal (15% = 0.15) and multiplying it by the original price:
$$\text{Discount Amount} = \$800 \times 0.15 = \$120$$

**Step 3: Subtract the discount from the original price**
Now, subtract the discount amount from the original price to find the final price:
$$\text{Final Price} = \$800 - \$120 = \$680$$

*(Alternatively, since a 15% discount means you are paying 85% of the original price, you can calculate it directly: $\$800 \times 0.85 = \$680$)*

**Final Answer:**
The final price of the laptop after applying the discount is **$680**.

# Structure-based prompt

````python
prompt = """
Explain Machine Learning.

Follow this exact structure:

1. Definition
- Give a simple definition.

2. Types
- Explain supervised learning.
- Explain unsupervised learning.
- Explain reinforcement learning.

3. How It Works
- Explain the basic process in simple language.

4. Real-World Applications
- Give 3 examples.


5. Advantages
- give 3 advantages


6. disAdvantages
- give 3 disadvantages

7. Conclusion
- Give a short conclusion.

Use simple English and beginner-friendly explanations."""

response = llm.models. generate_content(model="gemini-3.5-flash-lite",
contents=prompt)

print(response.text)
````

1. Definition
- Machine Learning is a branch of artificial intelligence (AI) that allows computers to learn from data and improve their performance over time without being explicitly programmed for every single task.

2. Types
- **Supervised learning:** The computer is trained using labeled data, meaning the input comes with the correct answer. It learns by comparing its predictions to the right answers and adjusting until it gets them right (like a student learning with an answer key).
- **Unsupervised learning:** The computer is given unlabeled data and must figure out patterns, similarities, or groupings on its own, without any guidance or correct answers provided.
- **Reinforcement learning:** The computer learns by trial and error. It takes actions in an environment and receives rewards for good actions and penalties for bad ones, maximizing its rewards over time (like training a pet with treats).

3. How It Works
- First, you collect a large amount of data. Second, you feed this data into a computer algorithm (a set of rules). Third, the algorithm analyzes the data to find patterns and make predictions. Finally, as the system gets feedback on its predictions, it automatically updates itself to become more accurate.

4. Real-World Applications
- Recommendation systems (like Netflix or YouTube suggesting what to watch next).
- Email spam filters (automatically sorting unwanted emails into the spam folder).
- Voice assistants (like Siri or Alexa understanding and responding to human speech).

5. Advantages
- It can handle huge amounts of complex data quickly.
- It improves automatically over time as it receives more data.
- It automates repetitive tasks, saving time and reducing human error.

6. Disadvantages
- It requires a lot of data and computing power to work well.
- It can make mistakes or inherit biases if the training data is flawed or unfair.
- It can be a "black box," meaning it is sometimes hard to understand how the computer reached a specific decision.

7. Conclusion
- Machine Learning is a powerful tool that powers many of the smart technologies we use every day, changing how we work, live, and solve complex problems.
